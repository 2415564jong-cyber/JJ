```python
"""
Program kasir restoran
Dibuat oleh: Jong Jun Lee 12-7/15
"""
menu_makanan = [
    ["Nasi Goreng", 15000],
    ["Mie Goreng", 12000],
    ["Ayam Goreng", 18000],
    ["Es Teh Manis", 5000],
    ["Es Jeruk", 6000],
]

file_nota = "nota_pembayaran.txt"

def tampilkan_daftar_menu():
    print("\n=== DAFTAR MENU RESTORAN ===")
    for index, item in enumerate(menu_makanan, start=1):
        print(f"{index}. {item[0]:<15} - Rp{item[1]:,.0f}")


def input_angka_positif(prompt):
    while True:
        try:
            nilai = int(input(prompt))
            if nilai <= 0:
                print("Jumlah harus lebih dari 0.")
                continue
            return nilai
        except ValueError:
            print("Input tidak valid! Harap masukkan angka bulat.")


def simpan_nota(pesanan, total_harga, bayar, kembalian):
    try:
        with open(file_nota, "w") as f:
            f.write("=== NOTA PEMBAYARAN RESTORAN ===\n\n")
            for item in pesanan:
                f.write(f"- {item['nama']} x{item['jumlah']} : Rp{item['subtotal']:,}\n")
            f.write("-" * 35 + "\n")
            f.write(f"Total     : Rp{total_harga:,}\n")
            f.write(f"Bayar     : Rp{bayar:,}\n")
            f.write(f"Kembalian : Rp{kembalian:,}\n")
        print(f"\nNota berhasil dicetak ke file '{file_nota}'.")
    except OSError as e:
        print(f"Gagal menyimpan nota: {e}")


def proses_pembayaran(pesanan):
    if not pesanan:
        print("\nBelum ada pesanan yang ditambahkan.")
        return False

    total_harga = sum(item["subtotal"] for item in pesanan)
    print(f"\nTotal Pembayaran: Rp{total_harga:,.0f}")

    while True:
        try:
            bayar = float(input("Masukkan uang pembayaran: Rp"))
            if bayar < total_harga:
                print(f"Uang Anda kurang Rp{total_harga - bayar:,.0f}.")
            else:
                kembalian = bayar - total_harga
                print(f"Pembayaran berhasil! Kembalian: Rp{kembalian:,.0f}")
                simpan_nota(pesanan, total_harga, bayar, kembalian)
                return True
        except ValueError:
            print("Input tidak valid! Masukkan nominal angka.")


def main():
    pesanan = []

    while True:
        print("\n===== SISTEM KASIR RESTORAN =====")
        print("1. Lihat menu dan tambah pesanan")
        print("2. Lihat pesanan saat ini")
        print("3. Bayar dan cetak nota")
        print("0. Keluar")

        pilihan = input("Pilih menu (0-3): ").strip()

        if pilihan == "1":
            tampilkan_daftar_menu()
            try:
                nomor_menu = int(input("Pilih nomor menu: "))
                if 1 <= nomor_menu <= len(menu_makanan):
                    item = MENU_MAKANAN[nomor_menu - 1]
                    jumlah = input_angka_positif("Masukkan jumlah porsi: ")
                    subtotal = item[1] * jumlah

                    pesanan.append({
                        "nama": item[0],
                        "harga": item[1],
                        "jumlah": jumlah,
                        "subtotal": subtotal,
                    })
                    print(f"Berhasil menambahkan {jumlah} {item[0]}.")
                else:
                    print("Nomor menu tidak tersedia.")
            except ValueError:
                print("Masukkan nomor menu berupa angka.")

        elif pilihan == "2":
            print("\n--- Daftar Pesanan Saat Ini ---")
            if not pesanan:
                print("Belum ada pesanan.")
            else:
                for idx, item in enumerate(pesanan, start=1):
                    print(f"{idx}. {item['nama']} ({item['jumlah']}x) - Rp{item['subtotal']:,.0f}")

        elif pilihan == "3":
            if proses_pembayaran(pesanan):
                pesanan.clear()

        elif pilihan == "0":
            print("Terima kasih telah menggunakan sistem kasir!")
            break
        else:
            print("Pilihan menu tidak dikenal. Coba lagi.")


if __name__ == "__main__":
    main()
```
